```python
# SISTEM PENCATATAN HASIL PANEN

def hitung_total_panen(jumlah_kg, harga_per_kg):
    return jumlah_kg * harga_per_kg


def hitung_diskon(total, persen_diskon):
    return total * persen_diskon / 100


# Data hasil panen
jumlah_panen = 100
harga_per_kg = 8000
persen_diskon = 10

# Menghitung total hasil panen
total = hitung_total_panen(jumlah_panen, harga_per_kg)

# Menghitung diskon
diskon = hitung_diskon(total, persen_diskon)

# Menghitung total setelah diskon
total_setelah_diskon = total - diskon

# Menampilkan hasil
print("SISTEM PENCATATAN HASIL PANEN")
print("Jumlah panen         :", jumlah_panen, "kg")
print("Harga per kg         : Rp", harga_per_kg)
print("Total hasil panen    : Rp", total)
print("Diskon", persen_diskon, "%           : Rp", diskon)
print("Total setelah diskon : Rp", total_setelah_diskon)
```
